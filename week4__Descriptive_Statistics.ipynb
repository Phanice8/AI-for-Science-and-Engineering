```python
#Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Load the dataset
df = pd.read_csv('/content/ai4i+2020+predictive+maintenance+dataset (3).zip') # Change this path if if your file has a different name

# Check the data
print("Dataset shape:", df.shape)
df.head()



# Select 4 numerical variables relevant to the project
variables = ['Air temperature [K]', 'Process temperature [K]',
             'Rotational speed [rpm]', 'Torque [Nm]']

# Calculate mean, median, range, variance, and standard deviation for each variable
for col in variables:
    print(f"\n========== Statistics for: {col} ==========")
    print(f"Mean: {df[col].mean():.2f}")
    print(f"Median: {df[col].median():.2f}")
    print(f"Range: {df[col].max() - df[col].min():.2f}")
    print(f"Variance: {df[col].var():.2f}")
    print(f"Standard Deviation: {df[col].std():.2f}")


    # Select two variables
var1 = 'Air temperature [K]'
var2 = 'Process temperature [K]'

# Covariance
covariance = df[[var1, var2]].cov().iloc[0, 1]
print(f"Covariance between {var1} and {var2}: {covariance:.4f}")

# Correlation
correlation = df[var1].corr(df[var2])
print(f"Correlation between {var1} and {var2}: {correlation:.4f}")


df = pd.read_csv('/content/ai4i+2020+predictive+maintenance+dataset (3).zip')

# Select two variables
var1 = 'Air temperature [K]'
var2 = 'Process temperature [K]'

# Covariance
covariance = df[[var1, var2]].cov().iloc[0, 1]
print(f"Covariance between {var1} and {var2}: {covariance:.4f}")

# Correlation
correlation = df[var1].corr(df[var2])
print(f"Correlation between {var1} and {var2}: {correlation:.4f}")


# Create a simple scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x=var1, y=var2, alpha=0.5)
plt.title(f'Relationship between {var1} and {var2}')
plt.xlabel(var1)
plt.ylabel(var2)
plt.grid(True, alpha=0.3)
plt.show()


### Explanation of the Statistics

##What do the statistics tell you about your data?**
##The mean and median of Air temperature and Process temperature are close to each other, which suggests the data is fairly symmetric. The range, variance, and standard deviation show how spread out the values are. These statistics help us understand the central tendency and variability of the sensor readings.

##What does the correlation tell you about the relationship between the two variables?
##The correlation between Air temperature and Process temperature is very high (close to 1). This means there is a strong positive linear relationship — when Air temperature increases, Process temperature also tends to increase.


##Does correlation necessarily mean that one variable causes the other?
##No. Correlation does not imply causation. Even though the two temperatures are strongly correlated, we cannot conclude that one causes the other. Both may be influenced by other factors such as the overall operating environment or machine settings.


### Simple Mathematical Expression

##We can represent the idea as:

##Input → Function/Model → Output**

##Example using our variables:

#\[\text{Air temperature [K]} \;\xrightarrow{\; f(x) \;} \; \text{Process temperature [K]}
#\]

#Or more generally:

#\[
#x \;\xrightarrow{\; f \;} \; y
#\]

Where:
# \( x \) = Air temperature [K] (Input)
#\( f \) = the relationship or model
# \( y \) = Process temperature [K] (Output)






```
